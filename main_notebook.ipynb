```python
import pandas as pd
from dataset_loader import HFDatasetLoader
from dataset_summarizer import DatasetSummarizer
```

```python
loader = HFDatasetLoader(dataset_name="scikit-learn/iris", split="train")
df_raw = loader.fetch_as_dataframe()
df_raw.head()
```

```python
# Transformación: Filtrar por SepalLengthCm > 5.0 y renombrar columnas al español
df_transformed = df_raw[df_raw["SepalLengthCm"] > 5.0].copy()

df_transformed = df_transformed.rename(
    columns={
        "Id": "id",
        "SepalLengthCm": "largo_sepalo",
        "SepalWidthCm": "ancho_sepalo",
        "PetalLengthCm": "largo_petalo",
        "PetalWidthCm": "ancho_petalo",
        "Species": "especie",
    }
)

df_transformed.head()
```

```python
summarizer = DatasetSummarizer(
    dataframe=df_transformed, output_path="resumen_dataset.txt"
)
summarizer.generate_summary()
```
